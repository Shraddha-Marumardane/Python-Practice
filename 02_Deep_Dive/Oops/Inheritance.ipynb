# SINGLE INHERITANCE

```python
class ParentClass:
    def Parent_Class_method(self):
        return " Parent class method  I have 50 acres of land"

class ChildClass(ParentClass):
    def ChildClass_method(self):
        return "Child class method I have Bike"

Parent_obj_01 = ParentClass()
Parent_obj_01.Parent_Class_method()
```

```python
Child_obj_01 = ChildClass()
print(Child_obj_01.ChildClass_method())
Child_obj_01.Parent_Class_method()
```

# Multi Level Inheritance

```python
class GrandParentsClass:
    def GrandParents_Method(self):
        return "GP_Method I have 500 Acres of Land"

class ParentClass(GrandParentsClass):
    def Parent_Class_method(self):
        return "Parent Class method I have 50 Acres of Land"

class ChildClass(ParentClass):
    def ChildClass_method(self):
        return "Child Class Method I have Bike"
GP_Obj_01 = GrandParentsClass()
GP_Obj_01.GrandParents_Method()
```

```python
Parent_Obj_01 = ParentClass()
print(Parent_Obj_01.Parent_Class_method())
print(Parent_Obj_01.GrandParents_Method())
```

```python
Child_Obj_01 = ChildClass()
print(Child_Obj_01.ChildClass_method())
print(Child_Obj_01.Parent_Class_method())
print(Child_Obj_01.GrandParents_Method())
```

# Single Inheritance

```python
class Company:
    def Company_Class_method(self):
        return "Company provides a laptop"

class Employee(Company):
    def Employee_Class_method(self):
        return "Employee works on projects"

Employee_obj_01 = Employee()

print(Employee_obj_01.Company_Class_method())
print(Employee_obj_01.Employee_Class_method())
```

# Multilevel Inheritance

```python
class CompanyClass:
    def Company_Method(self):
        return "Company Method: We provide laptops"

class EmployeeClass(CompanyClass):
    def Employee_Method(self):
        return "Employee Method: I work on projects"

class ManagerClass(EmployeeClass):
    def Manager_Method(self):
        return "Manager Method: I manage a team"


Company_Obj_01 = CompanyClass()
print(Company_Obj_01.Company_Method())

Employee_Obj_01 = EmployeeClass()
print(Employee_Obj_01.Employee_Method())
print(Employee_Obj_01.Company_Method())

Manager_Obj_01 = ManagerClass()
print(Manager_Obj_01.Manager_Method())
print(Manager_Obj_01.Employee_Method())
print(Manager_Obj_01.Company_Method())
```

# Multiple Inheritance

```python
class CompanyClass:
    def Company_Method(self):
        return "Company Method: We provide laptops"

class DepartmentClass:
    def Department_Method(self):
        return "Department Method: We provide training"

class EmployeeClass(CompanyClass, DepartmentClass):
    def Employee_Method(self):
        return "Employee Method: I work on projects"


Employee_Obj_01 = EmployeeClass()

print(Employee_Obj_01.Company_Method())
print(Employee_Obj_01.Department_Method())
print(Employee_Obj_01.Employee_Method())
```
