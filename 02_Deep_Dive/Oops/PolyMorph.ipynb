# Method Overloading

```python
# with overloading

class myClass:
    def sum_Overloaded(self,a=None,b=None,c=None,d=None):
        if a != None and b != None and c != None and d != None:
            return a+b+c+d
        if a != None and b != None and c != None:
            return a+b+c
        elif a != None and b != None:
            return a+b
        else:
            return "Pls pass atleast two values for sum"


my_Obj = myClass()
my_Obj.sum_Overloaded(1,2,3,4)
```

```python
# Simple method without overloading

class myClass:
    def sum_4(self,a,b,c,d):
        return a+b+c+d
    def sum_3(self,a,b,c):
        return a+b+c
    def sum_2(self,a,b,):
        return a+b

my_Obj = myClass()
my_Obj.sum_4(1,2,3,4)
my_Obj.sum_3(1,2,3)
my_Obj.sum_2(1,2)

```

# Method Overriding with Super keyword

```python
class Animal:
    def speak(self):
        print("The animal makes a sound")

class Dog(Animal):
    def speak(self):
        super().speak()
        print("The dog barks")

# create an instance of dog and call the speak method

dog = Dog()
dog.speak()

```

```python
# Use of super keyword

class p_c_interest:
    def interest(self):
        print(0.011)

class c_c_home_interest(p_c_interest):
    def interest(self):
        super().interest()
        print(0.08)

Home_obj =c_c_home_interest()
Home_obj.interest()
```
