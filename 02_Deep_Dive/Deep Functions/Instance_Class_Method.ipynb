```python
class Dog():
    def __init__(self):
        self.dog_Age = 14

dog_1 = Dog()
dog_1.dog_Age
```

```python
class Dog():
    def __init__(self,dog_Age):
        self.dog_Age = dog_Age          #Instance Variable

    def abc(self):                      #Instance Method
        print(self.dog_Age)

    def my_dog_details(self,colour):
        print(f"My Dog has {colour} color")

dog_1 = Dog(14)
dog_1.abc()
dog_1.my_dog_details("Brown")
```

```python
class Banking:
    Int_Rate = 0.08      # Class Variable

    @classmethod
    def change_int_Rate(cls):
        cls.Int_Rate = cls.Int_Rate + (cls.Int_Rate*0.02)
        print(cls.Int_Rate)

Banking.Int_Rate
Banking.change_int_Rate()
```
