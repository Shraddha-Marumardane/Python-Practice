```python
class Student:
    def __init__(self):
        self.name = "Shraddha"
        self._age = 22
        self.__marks = 90

    def display(self):
        print(self.name)                      #Public
        print(self._age)                      #Protected
        print(self.__marks)                   #Private

obj = Student()
obj.display()
```

```python
# Outside the class
class Student:
    def __init__(self):
        self.name = "Rahul"    # Public
        self._age = 20         # Protected
        self.__marks = 90      # Private

obj = Student()

# Access outside the class
print(obj.name)       # Public - Allowed
print(obj._age)       # Protected - Allowed
print(obj.__marks)    # Private - Error
```

```python
# Inside child class Inheritance
class ParentClass:                       # Parent Class
    def __init__(self):
        self.name = "Rahul"          # Public
        self._age = 20               # Protected
        self.__marks = 90            # Private

class ChildClass(ParentClass):      # Child Class
    def display(self):
        print(self.name)             # Public - Allowed
        print(self._age)             # Protected - Allowed
        print(self.__marks)          # Private - Error

Child_Obj = ChildClass()
Child_Obj.display()
```

```python
class Student:
    def __init__(self):
        self.name = "Rahul"    # Public
        self._age = 20         # Protected
        self.__marks = 90      # Private


class Teacher:
    def display(self):
        student_Obj = Student() 
        print(student_Obj.name)
        print(student_Obj._age)
        print(student_Obj.__marks)

tech_Obj = Teacher()
tech_Obj.display()
```

```python
# Within the class

class Student:
    def display_public(self):
        print("Public Method")

    def _display_protected(self):
        print("Protected Method")

    def _display_private(self):
        print("Private Method")

    def show(self):
        self.display_public()
        self._display_protected()
        self._display_private()

obj = Student()
obj.show()
```
