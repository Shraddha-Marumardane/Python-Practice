```python
# Home work
# Find the maching values with the index Number

col_01 = ["Red", "Yellow", "LightGreen"]
col_02 = ["Black", "White", "LightGreen"]
col_03 = ["Black", "LightGreen", "White", "Pink"]

for i in range(len(col_01)):
    for j in range(len(col_02)):
        for k in range(len(col_03)):
            if col_01[i] == col_02[j] == col_03[k]:
                print("Matching Value:", col_01[i])
                print("col_01 Index:", i)
                print("col_02 Index:", j)
                print("col_03 Index:", k)
```
