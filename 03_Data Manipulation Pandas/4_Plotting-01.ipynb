```python
import pandas as pd

filepath = r"C:\Users\shrad\Downloads\Airline1 (1).csv"
df = pd.read_csv(filepath)
df.head(2)
```

```python
import matplotlib.pyplot as plt

font = 15
plt.Figure(figsize = (7,2))
plt.bar(df['Airline'],df['Price'])
plt.xticks(rotation = 45)
plt.xlabel("Airline",fontsize = font)
plt.ylabel("Price", fontsize = font)
plt.title("Airline vs Price", fontsize = font)
plt.show()
```

# SEABORN

```python
import seaborn as sns

bp = sns.barplot(x = df['Airline'], y = df['Price'])
plt.xticks(rotation  = 45)
plt.show()
```

# Count plot

```python
# 
import seaborn as sns
plt.figure(figsize=(10,5))

ax = sns.countplot(x=df['Airline'],hue=df['Source'],width=.5)

ax.bar_label(ax.containers[0], fmt='%d', fontsize=8)
ax.bar_label(ax.containers[1], fmt='%d', fontsize=8)

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,4))
ax=sns.countplot(x=df['Total_Stops'],width=0.5)
ax.bar_label(ax.containers[0],fmt='%d',fontsize=8)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10,4))
ax=sns.countplot(x=df['Source'],width=0.5)
ax.bar_label(ax.containers[0],fmt='%d',fontsize=8)
plt.xticks(rotation=90)
plt.show()
```

# Line plot

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 5))

sns.lineplot( data=df, x='Airline', y='Price', marker='o')

plt.title('Price by Airline')
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize = (10,3))
sns.lineplot( x=df['Airline'], y=df['Price'], marker='o')

plt.xticks(rotation=90)
plt.title("Airline VS Price")
plt.show()
```

```python
plt.figure(figsize = (10,3))
sns.lineplot( x=df['Date_of_Journey'], y=df['Price'], marker='o')

plt.xticks(rotation=90)
plt.title("Airline VS Price")
plt.show()

```
