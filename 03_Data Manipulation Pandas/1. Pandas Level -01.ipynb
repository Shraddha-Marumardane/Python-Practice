```python
df = pd.read_csv(r"C:\Users\shrad\Downloads\Airline1.csv")
df.isna().sum()
df['Source'].value_counts()
Source_Value = df['Source'].mode()
Source_Value
df['Source'] = df['Source'].fillna(Source_Value[0])
df.isna().sum()
```

```python
dest_Value = df['Destination'].mode()

df['Destination'] = df['Destination'].fillna(dest_Value[0])
df.isna().sum()
```

```python
Mean_Price = df['Price'].mean()

df['Price'] = df['Price'].fillna(Mean_Price)
df.isna().sum()
```

```python
# Fill missing Price values with mean
Mean_Price = df['Price'].mean()
df['Price'] = df['Price'].fillna(Mean_Price)

# Fill missing Route values with 0
df['Route'] = df['Route'].fillna(0)

# Fill missing Total_Stops values with 0
df['Total_Stops'] = df['Total_Stops'].fillna(0)

# Check missing values
df.isna().sum()
```
