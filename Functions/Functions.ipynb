```python
# NO ARGUMENTS

def combineNames():
    FullName = "RAJESH" + " " + "Singh"
    print(FullName)

combineNames()

```

```python
# WITH ARGUMENTS

def printSquares(list_01):
    NewDict_12=" "
    NewDict_12 ={}
    for i in list_01:
        NewDict_12[i] = i*i
    print(NewDict_12)  
```

```python
list_03 = [10,20,30]
printSquares(list_03)
```

```python
def combineNames(Firstname,LastName):
    FullName = Firstname+" "+LastName
    print(FullName)
```

```python
# With Position Arguments
combineNames("Sunil","Patil")
```

```python
# With Keyword Arguments
combineNames(Firstname="Sunil",LastName="Patil")
```

```python
Firstname = "Sunil"
LastName = "Singh"
combineNames(Firstname,LastName)
```

```python
# Lambda Function

newFunc = lambda a : a +10

# name func lambda argume: expression 

newFunc(4)
```

```python
# Dict Comprehension
list_02 = [1, 2, 3]

def PrintSquares(list_01):
    NewDict_12 = {}
    for i in list_01:
        NewDict_12[i] = i * i
    print(NewDict_12)

PrintSquares(list_02)
```

```python
# Update the dict

Dict_01 = {"name": "Rajesh", "Age": 24}
Dict_02 = {"Pin": 444001, "Area": "Sanghavi"}

Dict_01.update(Dict_02)

print(Dict_01)
```

```python
# save values in new list which are less than 5

list_01 = [12,2,10,11,5,3]
def lessthanfive(list_01):
    new_list = []
    for i in list_01:
        if i < 5:
            new_list.append(i)
    print(new_list)

lessthanfive(list_01)
```
